```python
import pandas as pd  # Pandas untuk manipulasi dan analisis data
pd.options.mode.chained_assignment = None  # Menonaktifkan peringatan chaining
import numpy as np  # NumPy untuk komputasi numerik
```

# Srcapping Dataset

```python

from google_play_scraper import app, reviews_all, Sort # Import Sort, reviews_all, dan app dari google_play_scraper

# ID aplikasi
app_ids = ['com.telkom.tracencare']
all_reviews = []

for app_id in app_ids:
    print(f"Scrapping ulasan untuk aplikasi: {app_id}")
    reviews = reviews_all(
        app_id,
        lang='id',                # Bahasa ulasan (default: 'en')
        country='id',             # Negara (default: 'us')
        sort=Sort.MOST_RELEVANT,  # Urutan ulasan (default: Sort.MOST_RELEVANT)
    )
    all_reviews.extend(reviews)

scrapreview = all_reviews
```

# Loading Dataset

```python
app_reviews_df = pd.DataFrame(scrapreview)
app_reviews_df.shape
app_reviews_df.head()
app_reviews_df.to_csv('ulasan_aplikasi.csv', index=False)
```

```python
# Membuat DataFrame dari hasil scrapreview
app_reviews_df = pd.DataFrame(scrapreview)

# Menghitung jumlah baris dan kolom dalam DataFrame
jumlah_ulasan, jumlah_kolom = app_reviews_df.shape
```

```python
# Membuat DataFrame baru (clean_df) dengan menghapus baris yang memiliki nilai yang hilang (NaN) dari app_reviews_df
clean_df = app_reviews_df.dropna()
```

```python
# Menampilkan informasi tentang DataFrame clean_df setelah membersihkan nilai yang hilang (NaN)
clean_df.info()
```

```python
# Menghapus baris duplikat dari DataFrame clean_df
clean_df = clean_df.drop_duplicates()

# Menghitung jumlah baris dan kolom dalam DataFrame clean_df setelah menghapus duplikat
jumlah_ulasan_setelah_hapus_duplikat, jumlah_kolom_setelah_hapus_duplikat = clean_df.shape
```

```python
# Mengambil sampel acak dengan 30000 baris karna data hasil scrap terlalu besar bagi saya
clean_df = clean_df.sample(n=30000, random_state=42)
```

```python
# Menyimpan ulasan yang sudah dibersihkan dalam file CSV
clean_df.to_csv('ulasan_aplikasii.csv', index=False)
```
